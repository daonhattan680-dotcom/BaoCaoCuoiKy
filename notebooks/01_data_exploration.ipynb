KIỂM TRA TỔNG QUAN DỮ LIỆU

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Đọc dữ liệu
df = pd.read_csv(
    "../data/AirQualityUCI.csv",
    sep=";",
    decimal=","
)


print("5 dòng dữ liệu đầu tiên:")
display(df.head())


print("Kích thước dữ liệu:")
print("Số dòng:", df.shape[0])
print("Số cột:", df.shape[1])


print("\nDanh sách các cột:")
print(df.columns.tolist())


print("\nThông tin dữ liệu:")
df.info()
```

Bộ dữ liệu Air Quality ban đầu có 9.471 dòng và 17 cột. Qua quá trình kiểm tra, có 9.357 dòng chứa dữ liệu và 114 dòng trống ở cuối file. Trong đó, hai cột Unnamed: 15 và Unnamed: 16 không chứa giá trị nên được xác định là các cột thừa. Dữ liệu gồm các thông tin về nồng độ khí ô nhiễm, chỉ số cảm biến và điều kiện môi trường như nhiệt độ, độ ẩm. Hai thuộc tính Date và Time hiện đang ở dạng chuỗi, cần được chuyển đổi để phục vụ phân tích theo thời gian.

KIỂM TRA DỮ LIỆU THIẾU

```python
# Loại bỏ các dòng và cột trống
df = df.dropna(axis=0, how="all")
df = df.dropna(axis=1, how="all")

# Kiểm tra kích thước sau khi loại bỏ dữ liệu trống
print("Kích thước dữ liệu sau khi loại bỏ dòng/cột trống:")
print("Số dòng:", df.shape[0])
print("Số cột:", df.shape[1])

# Kiểm tra số lượng giá trị NaN
print("\nSố lượng giá trị thiếu (NaN) theo từng cột:")
print(df.isnull().sum())

# Kiểm tra số lượng giá trị -200
print("\nSố lượng giá trị -200 theo từng cột:")
numeric_df = df.select_dtypes(include=np.number)
print((numeric_df == -200).sum())

# Tổng số giá trị -200
print("\nTổng số giá trị -200:")
print((numeric_df == -200).sum().sum())
```

Sau khi loại bỏ các dòng và cột hoàn toàn trống, bộ dữ liệu còn 9.357 dòng và 15 cột. Kết quả kiểm tra cho thấy không có giá trị NaN. Tuy nhiên, dữ liệu có tổng cộng 16.701 giá trị -200. Đây là ký hiệu đại diện cho dữ liệu thiếu trong bộ dữ liệu Air Quality, không phải giá trị đo thực tế. Vì vậy, cần xử lý các giá trị này trước khi tiến hành huấn luyện mô hình Machine Learning.

```python
# Tạo bảng thống kê giá trị -200
missing_200 = (df.select_dtypes(include=np.number) == -200).sum()

# Tính tỷ lệ phần trăm
missing_percent = (missing_200 / len(df)) * 100

# Tạo bảng tổng hợp
missing_table = pd.DataFrame({
    "Số lượng thiếu": missing_200,
    "Tỷ lệ thiếu (%)": missing_percent.round(2)
})

# Sắp xếp từ nhiều đến ít
missing_table = missing_table.sort_values(
    by="Số lượng thiếu",
    ascending=False
)

display(missing_table)
```

Qua quá trình kiểm tra dữ liệu thiếu, có thể nhận thấy cột NMHC(GT) có tỷ lệ thiếu cao nhất, chiếm 90,23%. Các thuộc tính CO(GT), NO2(GT) và NOx(GT) có tỷ lệ thiếu khoảng 17,5% đến 18%. Trong khi đó, các thuộc tính cảm biến và môi trường còn lại có tỷ lệ thiếu khoảng 3,91%. Đặc biệt, biến mục tiêu NO2(GT) có 1.642 giá trị thiếu, cần được xem xét khi xây dựng tập dữ liệu huấn luyện. Vì vậy, cần có phương án xử lý dữ liệu thiếu phù hợp trước khi xây dựng mô hình Machine Learning.

THỐNG KÊ MÔ TẢ DỮ LIỆU

```python
# Chọn các cột số
numeric_df = df.select_dtypes(include=np.number)

# Tạm thời thay -200 bằng NaN để thống kê chính xác hơn
numeric_clean = numeric_df.replace(-200, np.nan)

# Thống kê mô tả
display(numeric_clean.describe().T)
```

Qua thống kê mô tả, biến mục tiêu NO2(GT) có giá trị trung bình khoảng 113,09, trung vị là 109, giá trị nhỏ nhất là 2 và lớn nhất là 340. Điều này cho thấy nồng độ NO2 có sự biến động trong tập dữ liệu. Bên cạnh đó, thuộc tính NMHC(GT) chỉ có 914 giá trị hợp lệ, thấp hơn đáng kể so với các thuộc tính còn lại. Các biến môi trường như nhiệt độ (T), độ ẩm (RH) và độ ẩm tuyệt đối (AH) cũng có sự thay đổi giữa các thời điểm. Những đặc điểm này cần được xem xét trong quá trình phân tích và lựa chọn đặc trưng cho mô hình.

```python
# Lấy dữ liệu NO2 hợp lệ
no2_data = df["NO2(GT)"].replace(-200, np.nan).dropna()

# Vẽ biểu đồ phân phối
plt.figure(figsize=(10, 5))

sns.histplot(
    no2_data,
    bins=30,
    kde=True
)

plt.title("Phân phối nồng độ NO2 trong không khí")
plt.xlabel("Nồng độ NO2 (µg/m³)")
plt.ylabel("Tần suất")

plt.show()
```

Dựa vào biểu đồ phân phối, nồng độ NO2 trong không khí có phân phối lệch phải. Đa số các giá trị tập trung ở mức trung bình thấp (khoảng 80 – 130 ug/m^3), trong khi đuôi phân phối kéo dài đến trên 300  ug/m^3, cho thấy sự xuất hiện của một số thời điểm có nồng độ chất ô nhiễm tăng cao đột biến. 

```python
# Tạo bản dữ liệu phục vụ phân tích thời gian
time_df = df.copy()

# Chuyển -200 thành NaN
time_df["NO2(GT)"] = time_df["NO2(GT)"].replace(-200, np.nan)

# Ghép ngày và giờ
time_df["DateTime"] = pd.to_datetime(
    time_df["Date"] + " " + time_df["Time"],
    format="%d/%m/%Y %H.%M.%S"
)

# Loại bỏ những dòng không có NO2
time_df = time_df.dropna(subset=["NO2(GT)"])

# Sắp xếp theo thời gian
time_df = time_df.sort_values("DateTime")

# Vẽ biểu đồ
plt.figure(figsize=(14, 5))

plt.plot(
    time_df["DateTime"],
    time_df["NO2(GT)"]
)

plt.title("Sự thay đổi nồng độ NO2 theo thời gian")
plt.xlabel("Thời gian")
plt.ylabel("Nồng độ NO2 (µg/m³)")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

plt.show()
```

Dựa trên biểu đồ chuỗi thời gian, nồng độ NO2 biến động mạnh và liên tục trong suốt giai đoạn quan sát. Đặc biệt, mức độ ô nhiễm có xu hướng tăng rõ rệt vào giai đoạn từ cuối năm 2004 đến đầu năm 2005, với nhiều thời điểm đạt đỉnh cao vượt mốc 300 ug/m^3. 

```python
# Chọn các cột cần phân tích
columns = [
    "NO2(GT)",
    "CO(GT)",
    "NOx(GT)",
    "C6H6(GT)",
    "T",
    "RH",
    "AH"
]

# Thay -200 bằng NaN
corr_data = df[columns].replace(-200, np.nan)

# Tính ma trận tương quan
corr_matrix = corr_data.corr()

display(corr_matrix)
```

NO2(GT) có tương quan dương mạnh nhất với NOx(GT) (0,763), tiếp theo là CO(GT) (0,683) và C6H6(GT) (0,614). Các biến AH, T và RH có tương quan âm với NO2(GT). Như vậy, NOx, CO và C6H6 có mối liên hệ đáng chú ý với nồng độ NO2.

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Ma trận tương quan giữa NO2 và các yếu tố")
plt.tight_layout()

plt.show()
```

Dựa trên ma trận tương quan, nồng độ CO2 có mối quan hệ tương quan thuận chặt chẽ với các chất ô nhiễm khác như NOx (r = 0.76), CO (r = 0.68) và C6H6 (r = 0.61), phản ánh chúng có chung nguồn phát thải. Trong khi đó, các yếu tố khí tượng như nhiệt độ và độ ẩm có hệ số tương quan thấp, cho thấy tác động trực tiếp của chúng đến nồng độ NO2 là không đáng kể. 

```python
plt.figure(figsize=(9, 5))

sns.boxplot(x=df["NO2(GT)"].replace(-200, np.nan))

plt.title("Boxplot nồng độ NO2")
plt.xlabel("Nồng độ NO2 (µg/m³)")

plt.show()
```
